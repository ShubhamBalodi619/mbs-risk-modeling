# Merging all origination and performance files

```python
# List all DataFrames you want to merge vertically
all_origination_dfs = [d14o, d15o, d16o, d17o, d18o, d19o, d20o, d21o, d22o, d23o, d24o]

# Concatenate them into a single master DataFrame
# ignore_index=True ensures the final DataFrame has a clean, continuous index
df_orig_master = pd.concat(all_origination_dfs, ignore_index=True)

print("Vertical merge complete. Final DataFrame is df_perf_master.")
print(f"Total rows in master performance file: {df_orig_master.shape[0]:,}")
df_orig_master.head()
```

```python
# List all DataFrames you want to merge vertically
all_performance_dfs = [d14, d15, d16, d17, d18, d19, d20, d21, d22, d23, d24]

# Concatenate them into a single master DataFrame
# ignore_index=True ensures the final DataFrame has a clean, continuous index
df_perf_master = pd.concat(all_performance_dfs, ignore_index=True)

print("Vertical merge complete. Final DataFrame is df_perf_master.")
print(f"Total rows in master performance file: {df_perf_master.shape[0]:,}")
print(df_perf_master.head())
```

# Joining origination and performance files

```python


# Ensure columns exist and clean names (this is crucial for merge success)
if 'Loan Sequence Number' in df_orig_master.columns:
    df_orig_master.columns = df_orig_master.columns.str.strip()
    df_orig_master.rename(columns={'Loan Sequence Number': 'Loan_ID'}, inplace=True)
elif 'Loan_ID' in df_orig_master.columns:
    df_orig_master.columns = df_orig_master.columns.str.strip()
    
if 'Loan Sequence Number' in df_perf_master.columns:
    df_perf_master.columns = df_perf_master.columns.str.strip()
    df_perf_master.rename(columns={'Loan Sequence Number': 'Loan_ID'}, inplace=True)
elif 'Loan_ID' in df_perf_master.columns:
    df_perf_master.columns = df_perf_master.columns.str.strip()


# --- 2. Perform the Horizontal Merge (Left Join) ---
# Target: df_perf_master (the large file with all monthly rows)
# Source: df_orig_master (the small file with unique loan data)
df_master_merged = df_perf_master.merge(
    df_orig_master, 
    on='Loan_ID', 
    how='left',
    suffixes=('_perf', '_orig') # Ensures clear naming if columns conflict
)

# --- 3. Verification ---
print("Horizontal Merge Complete.")
print(f"Final Master Data Shape (Rows, Columns): {df_master_merged.shape}")
print("\nVerifying merge success (static data must be repeated for each month):")

# Display a single loan's history to confirm the static data (Credit Score) is repeated
# and the dynamic data (Loan Age) changes.
verification_cols = ['Loan_ID', 'Loan Age', 'Monthly Reporting Period', 'Credit Score', 'Original UPB']
print(df_master_merged[verification_cols].head(10))
```

# Saving Final Merged File

```python
# Save the DataFrame to a CSV file
df_master_merged.to_csv(
    'final_master_loan_data.csv', 
    index=False,               # CRITICAL: Do NOT save the pandas row index as a column
    encoding='utf-8'           # Standard encoding for compatibility
)

print("File saved successfully as 'final_master_loan_data.csv' in the current directory. ✅")
```
